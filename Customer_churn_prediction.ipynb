```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
df.drop('customerID',axis='columns',inplace=True)
```

```python
df1 = df[df.TotalCharges!=' ']
df1.shape
```

```python
df1.dtypes
```

```python
df1.TotalCharges = pd.to_numeric(df1.TotalCharges)
```

```python
df1.dtypes
```

```python
tenure_churn_no= df1[df1.Churn == "No"].tenure
tenure_churn_yes= df1[df1.Churn == "Yes"].tenure
plt.hist([tenure_churn_yes,tenure_churn_no], color =['green', 'red'], label= ['Yes', 'No'])
plt.ylabel('No of Customers')
plt.xlabel('Tenure')
plt.legend()
```

```python
mc_churn_no= df1[df1.Churn == "No"].MonthlyCharges
mc_churn_yes= df1[df1.Churn == "Yes"].MonthlyCharges
plt.hist([mc_churn_yes,mc_churn_no], color =['green', 'red'], label= ['Yes', 'No'])
plt.ylabel('No of Customers')
plt.xlabel('Monthly charges')
plt.legend()
```

```python
def print_unique_col(df):
    for column in df.select_dtypes(include=['object', 'string']).columns:
        print(f"{column}: {df[column].unique()}")
```

```python
print_unique_col(df1)
```

```python
df1.replace('No internet service', 'No', inplace=True)
df1.replace('No phone service', 'No', inplace=True)

```

```python
print_unique_col(df1)
```

```python
yes_no_columns = [
    'Partner', 'Dependents', 'PhoneService', 'MultipleLines',
    'OnlineSecurity', 'OnlineBackup', 'DeviceProtection',
    'TechSupport', 'StreamingTV', 'StreamingMovies',
    'PaperlessBilling', 'Churn'
]

for col in yes_no_columns:
    df1[col] = df1[col].map({'Yes': 1, 'No': 0})
```

```python
for column in df1:
        print(f'{column} : {df1[column].unique()}')
```

```python
df1['gender'] = df1['gender'].map({'Female': 1, 'Male': 0})
```

```python
print(df1.dtypes)
```

```python
df2 = pd.get_dummies(
    df1,
    columns=['InternetService', 'Contract', 'PaymentMethod'],
    dtype='int'
)
```

```python
print(df2.select_dtypes(include=['object', 'string']).columns)
```

```python
df2.dtypes
```

```python
col_to_scale = ['tenure','MonthlyCharges','TotalCharges']

from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
df2[col_to_scale] = scaler.fit_transform(df2[col_to_scale])
```

```python
for column in df2:
        print(f'{column} : {df2[column].unique()}')
```

```python
print(df2.dtypes)
```

```python
print(df2.select_dtypes(include=['object', 'string']).columns)
```

```python
X = df2.drop('Churn', axis='columns')
y= df2['Churn']
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test, y_train, y_test = train_test_split(X,y,test_size= 0.2, random_state=5) 
```

```python
print(X_train.dtypes)
```

```python
print(X_train.select_dtypes(include=['object', 'string']).columns)
```

```python
print(type(X_train))
print(type(y_train))
```

```python
import tensorflow as tf
from tensorflow import keras

model = keras.Sequential([
    keras.layers.Dense(20, input_shape=(26,), activation='relu'),
    keras.layers.Dense(1,activation='sigmoid'),
])
model.compile(optimizer='adam', loss= 'binary_crossentropy', metrics= ['accuracy'])
model.fit(X_train,y_train, epochs=100)
```

```python
model.evaluate(X_test,y_test)
```

```python
yp = model.predict(X_test)
yp[:5]
```

```python
y_pred = []
for element in yp:
    if element > 0.5:
        y_pred.append(1)
    else:
        y_pred.append(0)
```

```python
y_pred[:10]
```

```python
y_test[:10]
```

```python
from sklearn.metrics import confusion_matrix , classification_report

print(classification_report(y_test,y_pred))
```
